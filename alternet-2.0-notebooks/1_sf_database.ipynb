### ENCODE

```python
en_raw = pd.read_excel('/Users/zihengdai/Desktop/Thesis/data/ENCODE/41586_2020_2077_MOESM3_ESM.xlsx', header=1)
en_raw
```

```python
old_name = en_raw.columns[0]
en_raw.rename(columns={old_name: 'RBP_Name'}, inplace=True)
en_filtered = en_raw[["RBP_Name", 'Splicing regulation', 'Spliceosome']].copy()
en_filtered
```

```python
en_sf = en_filtered[
    (en_filtered['Splicing regulation'].fillna(0) == 1) |
    (en_filtered['Spliceosome'].fillna(0) == 1)
]

en = en_sf['RBP_Name'].to_list()
en
```

### Union of all unique SFs 

```python
saf_set = set(saf)
sl_set = set(sl)
ks_set = set(ks)
en_set = set(en)

comprehensive_sfs_set = saf_set.union(sl_set, ks_set, en_set)
comprehensive_sfs = sorted(list(comprehensive_sfs_set))
comprehensive_sfs
```

```python
len(comprehensive_sfs)
```

```python
pd.DataFrame(comprehensive_sfs).to_csv("comprehensive_sfs.csv", index=False, header=False)
```

```python
final_df = pd.DataFrame(list(comprehensive_sfs), columns=['Splicing_Factor'])
final_df = final_df.sort_values('Splicing_Factor').reset_index(drop=True)

final_df['In_SplicingLore'] = final_df['Splicing_Factor'].isin(sl_set)
final_df['In_SpliceAidF'] = final_df['Splicing_Factor'].isin(saf_set)
final_df['In_KEGG'] = final_df['Splicing_Factor'].isin(ks_set)
final_df['In_ENCODE'] = final_df['Splicing_Factor'].isin(en_set)
```

```python
annotation_columns = ['In_SplicingLore', 'In_SpliceAidF', 'In_KEGG', 'In_ENCODE']
final_df['Source_Count'] = final_df[annotation_columns].sum(axis=1)

overlap_counts = final_df['Source_Count'].value_counts().sort_index(ascending=False)

high_confidence_sfs = final_df[final_df['Source_Count'] == 4]['Splicing_Factor'].tolist()
```
