```python
import os
import glob
import pandas as pd

directory_path = "/data/bionets/og86asub/alternet-project/alternet/data/rna_target"
all_files = glob.glob(os.path.join(directory_path, "*.csv"))

dfs = []
for file_path in all_files:
    gene_name = os.path.splitext(os.path.basename(file_path))[0]
    print(gene_name)
    df = pd.read_csv(file_path)
    df.columns = ['target', 'target_ensemble', 'type', 'protocol', 'target_expression', 'binding_site_records']
    df["source"] = gene_name
    dfs.append(df)

combined_df = pd.concat(dfs, ignore_index=True)

```

```python
import os.path as op
biomart_path = "biomart.txt"

data_path = "/data/bionets/og86asub/alternet-project/alternet/data"
biomart = pd.read_csv(op.join(data_path, biomart_path), sep='\t')
```

```python
mart_mapping = dict(zip(biomart['Gene name'], biomart['Gene stable ID']))
```

```python
combined_df['source_ensemble'] = combined_df.source.map(mart_mapping)
```

```python
combined_df.to_csv('/data/bionets/og86asub/alternet-project/alternet/data/splicefactor_targets.tsv', sep = '\t')
```
