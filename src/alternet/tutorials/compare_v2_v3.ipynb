```python
import pandas as pd
```

```python
v3_fa = pd.read_csv('/data/bionets/og86asub/alternet-project/results_v3/fully_as.tsv', sep = '\t')
v3_sa = pd.read_csv('/data/bionets/og86asub/alternet-project/results_v3/source_as.tsv', sep = '\t')
v3_canonical = pd.read_csv('/data/bionets/og86asub/alternet-project/results_v3/canonical.tsv', sep = '\t')
```

```python
v2_st = pd.read_csv('/data/bionets/og86asub/alternet-project/results_v2/fully_as.tsv', sep = '\t')
v2_sg = pd.read_csv('/data/bionets/og86asub/alternet-project/results_v2/canonical.tsv', sep = '\t')
```

```python
v3 = pd.concat([v3_fa, v3_sa, v3_canonical])
v3['ranke']
```

```python
v3 = v3.sort_values('importance', ascending=False)
v3['rank'] = v3['importance'].rank(ascending=False, method='min')
```

```python
v2 = pd.concat([v2_st, v2_sg])
v2 = v2.sort_values('importance', ascending=False)
v2['rank'] = v2['importance'].rank(ascending=False, method='min')
```

```python
m = v3.merge(v2, on =['TF', 'target'])
```

```python
m['difference'] = m['importance_x']-m['importance_y']
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Optional: Sort by rank to ensure a clean line plot
m = m.sort_values(by='rank_x')

# 2. Set a clean aesthetic theme
sns.set_theme(style="whitegrid")

# 3. Create the plot
# Use sns.lineplot if you want a continuous line connecting the points
ax = sns.lineplot(data=m.iloc[0:25000], x='rank_x', y='difference', marker='.', color='b')

# Alternatively, use sns.scatterplot if you only want the data points:
# ax = sns.scatterplot(data=df, x='rank', y='difference', color='b')

# 4. Customize the labels and title
ax.set_xlabel('Rank')
ax.set_ylabel('Difference')
ax.set_title('Difference vs. Rank')

# 5. Ensure layout is clean and save/show the figure
plt.tight_layout()
plt.savefig('difference_vs_rank.png', dpi=300)
```
