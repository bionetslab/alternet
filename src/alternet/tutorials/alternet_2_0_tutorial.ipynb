# AlterNet 2.0 Tutorial

This notebook demonstrates the AlterNet 2.0 pipeline for alternative splicing-aware
gene regulatory network inference.

## Overview

AlterNet 2.0 extends the original AlterNet pipeline with:

1. **Splicing Factor (SF) support** – builds a comprehensive SF database from
   SpliceAid-F, SplicingLore, KEGG, and ENCODE sources.
2. **Three-network inference** – infers canonical (gene-level), AS-aware source
   (transcript regulator), and fully AS-aware (transcript-to-transcript) networks.
3. **Edge categorization** – classifies edges by source and target resolution.
4. **Plausibility filtering** – removes implausible isoform-specific claims.
5. **Enrichment input** – generates ranked gene lists for g:Profiler ORA.
